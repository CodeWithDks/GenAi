```python
from langchain_core.tools import tool


@tool
def calculate(
    a: float,
    b: float,
    operation: str
) -> float:
    """
    Perform a basic mathematical calculation.

    Operations:
    - add
    - subtract
    - multiply
    - divide
    - percentage

    For percentage:
    calculates b percent of a.
    """

    if operation == "add":
        return a + b

    if operation == "subtract":
        return a - b

    if operation == "multiply":
        return a * b

    if operation == "divide":
        if b == 0:
            raise ValueError("Cannot divide by zero.")
        return a / b

    if operation == "percentage":
        return (a * b) / 100

    raise ValueError(
        "Unsupported operation."
    )
```

```python
result = calculate.invoke({
    "a": 10,
    "b": 5,
    "operation": "multiply"
})

print(result)
```

```python
calculate.invoke({
    "a": 800,
    "b": 25,
    "operation": "multiply"
})
```

```python
from langchain_openai import ChatOpenAI
from dotenv import  load_dotenv

load_dotenv()

llm = ChatOpenAI(
    model="gpt-4o-mini",
    temperature=0
)
```

```python
llm_with_tools = llm.bind_tools(
    [calculate]
)
```

```python
response = llm_with_tools.invoke(
    "how much 20 mangos and 30 apples"
)
```

```python
print(response)
print(response.tool_calls)
```

```python
# 1. Get the first tool call requested by the LLM
tool_call = response.tool_calls[0]
print(tool_call)
```

```python
# 2. Extract the arguments: {'a': 800, 'b': 25, 'operation': 'percentage'}
tool_args = tool_call["args"]
print(tool_args)
```

```python
# 3. Execute the tool with those arguments
result = calculate.invoke(tool_args)
```

```python
print("Actual Result:", result)
# Output: Actual Result: 200.0
```
